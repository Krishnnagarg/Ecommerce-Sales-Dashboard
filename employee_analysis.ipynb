```python
# step1:- Import Required Libraries
import pandas as pd
import numpy as np
import openpyxl
import pymysql
import sqlalchemy

print("All libraries working!")
```

```python
# step2:- Load DataSet

df = pd.read_excel(r"data/chapter11-Project-2.xlsx")
```

```python
# Step4 :- Check Missing Values
df.isnull().sum()
```

```python
# Step5 :- Check Duplicate Records

df.duplicated().sum()
```

```python
# Remove Duplicate Records

df.drop_duplicates(inplace=True) # inplace = true means original dataframe chanegs ho
```

```python
df.duplicated().sum()
```

```python
# Step6 :- Replace Invalid Values
# Replace Common invalid values to Nan
 
df.replace(["N/A","NULL",""],np.nan,inplace=True)
```

```python
df
```

```python
# Step7:- Remove Extra Spaces

df = df.apply(lambda x: x.str.strip() if x.dtype =='object' else x)  
```

```python
df
```

```python
#step8:- Standardize Text Format 
# Convert Text into Proper Case

df['Customer_Name'] = df['Customer_Name'].str.title()
df['City'] = df['City'].str.title()
df['State'] = df['State'].str.title()
```

```python
#Step9 :- Validate Email Addresses
# Keep Only Valid Email Records

df = df[df['Email'].str.contains('@',na=False)]
```

```python
#step10:- Convert Date COlumns
df['Order_Date'] = pd.to_datetime(df['Order_Date'],errors='coerce') #NAT
df["Delivery_Date"] =pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
#step11:- Convert Numeric Columns 

#df['Qty'] = pd.to_numeric(df['Qty'],errors="coerce")
#df['Unit_Price'] = pd.to_numeric(df['Unit_Price'],errors="coerce")


cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
#step12:- Remove Invalid Quantity
# Remove Records Having Zero and Negative Quantity

df = df[df["Qty"]>0]
```

```python
#step13 :- Fill Missing Values
df['Discount'] = df['Discount'].fillna(0)
df['Phone'] = df['Phone'].fillna("Unknown")
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
#step14:- Feature Enginnering
# Total Sales

df['Sales'] = df['Qty']*df['Unit_Price']
```

```python
df
```

```python
# Net Amount

df["Net_Amount"] = df["Sales"]-(df["Sales"]*df["Discount"]/100)
```

```python
# Profit (Example 20%)

df['Profit'] = df['Net_Amount']*0.20
```

```python
df
```

```python
# Order Month 

df["Month"] = df['Order_Date'].dt.month_name()
```

```python
df
```

```python
df["Year"] = df["Order_Date"].dt.year
```

```python
df
```

```python
# Show WeekDay

df["Weekday"] = df['Order_Date'].dt.day_name()
```

```python
df
```

```python
# step15 :- Final Data Validation

df.info()
```
